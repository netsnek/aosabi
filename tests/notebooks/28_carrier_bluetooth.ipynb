# 28 carrier bluetooth

Question: does Bluetooth carry GraphTP(Homa) on the L2CAP LE credit based channel from Linux,
Android, iOS and macOS clients, and on GATT, which every card with Bluetooth links unless it
switches it off?

The side under test is `aosabi-carrier-bluetooth`, its driver of Linux user space behind the
operation `serve` of `aosabi-devtool carrier`: the service of Aosabi registered with BlueZ over the
D-Bus client the crate writes itself, the PSM characteristic read with encryption, RX written
without response and TX notified, one Homa packet an ATT value, with one Homa endpoint of the server
role whose one peer is the central the data path serves and the GraphTP listener of `aosabi-core` in
front of the Yoga instance `prod`, or a responder of bytes where a row judges the Homa layer of the
link alone. It runs in guest A of `tests/guests/`, where `btvirt` of BlueZ 5.87 runs two virtual
controllers over `hci_vhci` that reach each other: the carrier serves on `hci0` and the central
`tests/harness/homa/bt_client.py` connects from `hci1` through the same `bluetoothd`, with the
Python Homa engine `tests/harness/homa/client.py` and the client of `tests/harness/homa/graphtp.py`,
all written apart from the crates.

The exit number of the emulated half, the plan's section The notebooks, H3: on the channel, 0
packets misframed when read in random chunks, a bind collision served on the number the kernel
allocated and published, a squatter on the other address form found at bind time, a request without
encryption refused with a security result, 1 disconnect failing every RPC `LinkDown`, 16 KiB within
10 percent of its packet time and 100 of 100 operations sent as text from the Linux central, and on
GATT LIMITS in the first 20 byte notification, 16 KiB within 10 percent of its packet time at an
ATT_MTU of 185, 1 dropped packet recovered through RESEND and 1 disconnect failing every RPC

answer behind 16 KiB on GATT against the 45 ms of the plan's section The budget, which names this
notebook. A crossed bound fails the change that crosses it and the owner decides whether the bound
moves. So where the parameter `wait_question` names the question, the row prints NOT MET beside its
measure and the 45 ms and does not fail the verdict, which is a visible state and not a pass, and
where the parameter is `None` the row fails the verdict when the wait exceeds the bound.

The central therefore sends the small request at a seeded moment spread uniformly over the download
of the large answer, whose span a calibration draw of the same phase measures, and times it from its
write of the request's packet at its event to the hand out of the answer, the mean of `wait_draws`
draws in both columns, GraphTP(Homa) and AoFrame alone. Starting at that write overstates the wait
of section 6.4 by the hop of the request to the driver, which the table per layer prints. The form
the record measured, the small request sent when the first packet of the large answer is handed out
and timed from its making with its uplink, prints beside it in both columns from `table_draws` draws
with no bound, since on the paced link it cannot fall below about two intervals whatever the carrier
does: a write made just after an event waits for the next one, and an answer cannot be handed out
before the event after that. The driver keeps a trace of its data path and the central stamps every
value and every write on the same realtime clock, so a table per layer splits the wait into the hop
of the request, the server and the queue below the engine, counts the packets of the large answer
below the engine when the driver sends the small answer, and marks each cause the record assumes
held or refuted.

The driver holds the packets below the engine to one connection event, the record's second direction
of the fix. Its data path counts every notification it hands BlueZ, control packets included, until
a pace of `packets_per_event` packets each connection interval of `interval_ms` retires it, and

a draw of section 6.4 that finds more than `tx_depth` packets of the large answer below the engine

## 1. The reader of the system bus against hostile messages

The D-Bus client of the crate is a new parser of bytes from the system bus in every Linux build with
Bluetooth, which the attack surface counts. The cell writes the messages the carrier meets with the
D-Bus writer of `tests/harness/homa/dbus.py`, written apart from the crate, a `WriteValue` with its
options, a `PropertiesChanged` of a device and a `GetManagedObjects` answer, and holds what the
crate reads to what the writer wrote. It then sends `hostile` messages the writer's messages become
when cut short, when a length or a signature is changed or when bytes are drawn at random, each on
its own, and counts the refusals, the messages the crate reads in full and any crash.

```python
rng = random.Random(seed)


def written(kind, fields, signature, body, serial):
    """A message written by the Python writer of the harness."""
    payload = dbus.Writer().write_all(signature, body) if signature else b""
    header_fields = [(code, dbus.Variant(sig, value)) for code, sig, value in fields]
    if signature:
        header_fields.append((dbus.SIGNATURE, dbus.Variant("g", signature)))
    head = dbus.Writer()
    head.buf += struct.pack("<cBBBII", b"l", kind, 0, 1, len(payload), serial)
    head.write("a(yv)", header_fields)
    head.pad(8)
    return bytes(head.buf) + payload


GOOD = [
    (written(dbus.METHOD_CALL, [(dbus.PATH, "o", "/org/aosabi/gatt/service0/char1"),
                                (dbus.INTERFACE, "s", "org.bluez.GattCharacteristic1"),
                                (dbus.MEMBER, "s", "WriteValue")],
             "aya{sv}", [b"\x01\x02\x03", {"device": dbus.Variant("o", "/org/bluez/hci0/dev_00_AA_01_01_00_24"),
                                            "mtu": dbus.Variant("q", 185), "type": dbus.Variant("s", "command")}], 7),
     {"kind": "MethodCall", "member": "WriteValue", "signature": "aya{sv}", "serial": 7}),
    (written(dbus.SIGNAL, [(dbus.PATH, "o", "/org/bluez/hci0/dev_00_AA_01_01_00_24"),
                           (dbus.INTERFACE, "s", "org.freedesktop.DBus.Properties"),
                           (dbus.MEMBER, "s", "PropertiesChanged")],
             "sa{sv}as", ["org.bluez.Device1", {"Connected": dbus.Variant("b", False)}, []], 8),
     {"kind": "Signal", "member": "PropertiesChanged", "signature": "sa{sv}as", "serial": 8}),
    (written(dbus.METHOD_RETURN, [(dbus.REPLY_SERIAL, "u", 3)], "a{oa{sa{sv}}}",
             [{"/org/bluez/hci0": {"org.bluez.Adapter1": {"Powered": dbus.Variant("b", True),
                                                           "Address": dbus.Variant("s", "00:AA:01:00:00:23")}}}], 9),
     {"kind": "MethodReturn", "signature": "a{oa{sa{sv}}}", "serial": 9}),
]
good_answers = devtool_batch("dbus-good", [{"id": i, "op": "dbus-decode", "hex": message.hex()}
                                          for i, (message, _) in enumerate(GOOD)])
GOOD_DIFFERENCES = []
for (message, want), answer in zip(GOOD, good_answers):
    read = (answer.get("messages") or [{}])[0]
    if any(read.get(key) != value for key, value in want.items()) or answer.get("offset") != len(message):
        GOOD_DIFFERENCES.append((want, answer))
print("messages of the writer read back: %d of %d" % (len(GOOD) - len(GOOD_DIFFERENCES), len(GOOD)))


def hostile_message(n):
    base = bytearray(GOOD[n % len(GOOD)][0])
    shape = n % 5
    if shape == 0:
        return bytes(base[:rng.randrange(1, len(base))])
    if shape == 1:
        at = rng.choice([4, 12])
        struct.pack_into("<I", base, at, rng.choice([0, 1, 0x7FFFFFFF, 0xFFFFFFFF, len(base) * 2]))
        return bytes(base)
    if shape == 2:
        for _ in range(rng.randint(1, 8)):
            base[rng.randrange(len(base))] = rng.randrange(256)
        return bytes(base)
    if shape == 3:
        signature = rng.choice(["a" * rng.randint(40, 200) + "y", "(" * 40 + ")" * 40, "a{", "a{vs}", "(y", "z", "{sv}"])
        return written(dbus.SIGNAL, [(dbus.PATH, "o", "/x"), (dbus.MEMBER, "s", "M"),
                                     (dbus.SIGNATURE, "g", signature)], "", [], 1) + bytes(rng.randint(0, 64))
    return b"l" + bytes(rng.randrange(256) for _ in range(rng.randint(15, 200)))


HOSTILE = [hostile_message(n) for n in range(hostile)]
hostile_answers = devtool_batch("dbus-hostile", [{"id": i, "op": "dbus-decode", "hex": message.hex()}
                                                for i, message in enumerate(HOSTILE)])
REFUSED = sum(1 for answer in hostile_answers if "refused" in answer)
INCOMPLETE = sum(1 for answer in hostile_answers if "incomplete" in answer)
READ_WHOLE = sum(1 for answer in hostile_answers if "refused" not in answer and "incomplete" not in answer)
HOSTILE_ANSWERED = len(hostile_answers)
print("hostile messages: %d sent, %d answered, %d refused, %d incomplete, %d read whole" % (
    len(HOSTILE), HOSTILE_ANSWERED, REFUSED, INCOMPLETE, READ_WHOLE))
```

## The guest and the controllers

Guest A boots from `target/guests/`, `btvirt -l2` runs as a transient unit of systemd and gives two
virtual controllers, `hci0` for the carrier and `hci1` for the central, and `bluetoothd` takes both.
Each phase writes `/etc/bluetooth/main.conf` with the `ATT_MTU` it runs at, restarts `bluetoothd`,
starts the driver on `hci0` and runs the central, which the cell writes into the exchange directory
from this notebook, in one command of the guest agent. The central registers a default agent with
the capability NoInputNoOutput, which accepts a pairing as Just Works and, since one `bluetoothd`
holds both controllers, answers the carrier's side too, reads the PSM characteristic first, whose
encryption makes BlueZ pair the two controllers, and subscribes to TX before its first write, as a
page does. A machine without KVM prints the guest half not run, which is never a pass.

```python
GUEST_REASON = hg.kvm_reason()
print("KVM:", GUEST_REASON or "available")
GUEST = None
CONTROLLERS = None
if GUEST_REASON is None:
    GUEST = hg.Guest("A").boot()
    print("guest A booted in %.1f s" % (GUEST.boot_seconds or 0.0))
    print(hg.placement(GUEST))
    CONTROLLERS = GUEST.run("modprobe hci_vhci && systemd-run --unit=btvirt /usr/local/bin/btvirt -l2 >/dev/null"
                            " && for i in $(seq 100); do [ -e /sys/class/bluetooth/hci1 ] && break; sleep 0.1; done;"
                            " ls /sys/class/bluetooth; /usr/libexec/bluetooth/bluetoothd --version").out.split()
    print("controllers and bluetoothd:", CONTROLLERS)

GUEST_PROGRAM = r'''
import collections, json, os, random, sys, time
sys.path.insert(0, "/mnt/tree/tests")
from harness.homa import bt_client, compact, dbus, client as hc, graphtp as g

PHASE, SPEC = sys.argv[1], json.loads(sys.argv[2])
RNG = random.Random(SPEC["seed"])
OUT = {}
UUIDS = {k: SPEC[k] for k in ("service", "psm", "rx", "tx")}
KANJI = "日本語の漢字をかな交じりで運ぶ東京大阪京都名古屋札幌福岡仙台広島神戸横浜"
ALPHABET = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789 -_.,:;"


class Seen:
    """A hook that records the incoming packets and drops one DATA packet on demand."""

    def __init__(self):
        self.first_in, self.drop_next_data, self.dropped, self.data_in = None, False, 0, 0

    def __call__(self, direction, packet):
        if direction != "in":
            return [packet]
        try:
            kind = compact.decode(packet)["packet"]["type"]
        except Exception:
            kind = "?"
        if self.first_in is None:
            self.first_in = (len(packet), kind)
        if kind == "DATA":
            self.data_in += 1
            if self.drop_next_data:
                self.drop_next_data = False
                self.dropped += 1
                return []
        return [packet]


def paired(paced=True):
    """Connect from hci1 to the carrier on hci0 with a Just Works agent, read
    the PSM, which pairs, subscribe to TX, and pace the link where `paced`."""
    bus = dbus.Bus()
    bt_client.register_agent(bus)
    address = dbus.get_property(bus, "org.bluez", "/org/bluez/hci0", "org.bluez.Adapter1", "Address")
    c = bt_client.GattCentral(address, UUIDS["service"], UUIDS["psm"], UUIDS["rx"], UUIDS["tx"], adapter="hci1", bus=bus)
    began = time.monotonic()
    c.connect(timeout=60)
    psm = c.read_psm()
    c.start_notify()
    if paced:
        c.pace(SPEC["interval_ms"] / 1e3, SPEC["per_event"])
    return c, psm, address, time.monotonic() - began


def central(hook, paced=True):
    c, psm, address, connect_s = paired(paced)
    mtu = c.att_mtu()
    packet = max(mtu - 3, 20)
    profile = hc.Profile(packet, 4 * packet, 8 * packet, 2.0, 18)
    transport = bt_client.GattTransport(c, packet, hook)
    engine = hc.Engine(transport, profile, hc.limits(profile, 131072, 8))
    return c, transport, engine, {"psm": psm, "att_mtu": mtu, "connect_s": connect_s, "address": address}


def drive(engine, rpcs, timeout):
    deadline = time.monotonic() + timeout
    while any(r.state in ("out", "in") for r in rpcs) and time.monotonic() < deadline:
        engine.poll(0.002)
        if engine.t.closed:
            engine.link_down()
            break


def disconnect_case(c, engine, rpcs_count, size):
    flying = [engine.request(RNG.randbytes(size)) for _ in range(rpcs_count)]
    for _ in range(3):
        engine.poll(0.002)
    c.bus.call("org.bluez", c.device, "org.bluez.Device1", "Disconnect")
    cut = time.monotonic()
    seconds = None
    while time.monotonic() - cut < 20:
        engine.poll(0.005)
        if engine.t.closed:
            engine.link_down()
        if all(r.state == "failed" for r in flying):
            seconds = time.monotonic() - cut
            break
    return {"rpcs": len(flying), "failed_link_down": sum(r.state == "failed" and r.why == "LinkDown" for r in flying),
            "seconds": seconds}


def phase_limits():
    seen = Seen()
    c, t, e, info = central(seen, paced=False)
    OUT["central"] = info
    rpc = e.request(b"limits first")
    drive(e, [rpc], 30)
    OUT["first_notification"] = {"length": seen.first_in[0] if seen.first_in else None,
                                 "type": seen.first_in[1] if seen.first_in else None,
                                 "echo_ok": rpc.state == "ok" and rpc.response == b"limits first"}
    OUT["disconnect"] = disconnect_case(c, e, 4, 4096)


def phase_transfer():
    seen = Seen()
    c, t, e, info = central(seen)
    OUT["central"] = info
    times, packets = [], []
    for _ in range(SPEC["transfers"]):
        before = seen.data_in
        began = time.monotonic()
        rpc = e.request(b"16k")
        drive(e, [rpc], 60)
        times.append(time.monotonic() - began)
        packets.append(seen.data_in - before)
        assert rpc.state == "ok" and len(rpc.response) == 16384, (rpc.state, rpc.why)
    OUT["transfer"] = {"seconds": times, "data_packets": packets}
    resends_before = e.stats["resends"]
    seen.drop_next_data = True
    began = time.monotonic()
    rpc = e.request(b"16k")
    drive(e, [rpc], 60)
    OUT["drop"] = {"dropped": seen.dropped, "ok": rpc.state == "ok" and len(rpc.response) == 16384,
                   "resends": e.stats["resends"] - resends_before, "seconds": time.monotonic() - began}
    OUT["disconnect"] = disconnect_case(c, e, 4, 16)


class Baseline:
    """The central of AoFrame alone: whole messages in prefixed frames cut to values of `packet`
    bytes, with the hand out of the value that completed the last message taken in `last_ns`."""

    def __init__(self, c, packet):
        self.c, self.packet, self.inbuf = c, packet, bytearray()
        self.counts = {"link_bytes_out": 0, "link_bytes_in": 0, "packets_in": 0}
        self.taken = 0
        self.ends = collections.deque()
        self.last_ns = None

    def send(self, message):
        frame = len(message).to_bytes(4, "big") + message
        self.counts["link_bytes_out"] += len(frame)
        for start in range(0, len(frame), self.packet):
            self.c.write(frame[start:start + self.packet])

    def poll(self, timeout):
        """Take the values handed out within `timeout`, each with the end of its bytes in the
        stream and its hand out."""
        for entry in self.c.notifications(timeout, stamped=True):
            if entry is None:
                raise ConnectionError("the link went down")
            self.counts["link_bytes_in"] += len(entry[2])
            self.counts["packets_in"] += 1
            self.inbuf += entry[2]
            self.ends.append((self.taken + len(self.inbuf), entry[1]))

    def take(self):
        """Return the next whole message or None, keeping the hand out of the value that
        completed it."""
        if len(self.inbuf) < 4:
            return None
        length = int.from_bytes(self.inbuf[:4], "big")
        if len(self.inbuf) < 4 + length:
            return None
        message = bytes(self.inbuf[4:4 + length])
        del self.inbuf[:4 + length]
        self.taken += 4 + length
        while self.ends[0][0] < self.taken:
            self.ends.popleft()
        self.last_ns = self.ends[0][1]
        if not self.inbuf:
            self.ends.clear()
        return message

    def receive(self, timeout=60.0):
        deadline = time.monotonic() + timeout
        while True:
            message = self.take()
            if message is not None:
                return message
            self.poll(0.002)
            if time.monotonic() > deadline:
                raise TimeoutError("no whole message")


def phase_baseline_transfer():
    c = paired()[0]
    base = Baseline(c, max(c.att_mtu() - 3, 20))
    base.send(b"warm")
    base.receive()
    times, packets = [], []
    for _ in range(SPEC["transfers"]):
        before = base.counts["packets_in"]
        began = time.monotonic()
        base.send(b"16k")
        message = base.receive()
        times.append(time.monotonic() - began)
        packets.append(base.counts["packets_in"] - before)
        assert len(message) == 16384
    OUT["transfer"] = {"seconds": times, "packets": packets, "att_mtu": c.att_mtu()}
    c.disconnect()


def echo_content(v):
    return json.dumps({"query": "query($v:String!){echo(value:$v)}", "variables": {"v": v}},
                      separators=(",", ":"), ensure_ascii=False).encode("utf-8")


def echoed(outcome):
    try:
        return json.loads(outcome["body"])["data"]["echo"] if outcome.get("answer") == "result" else None
    except (ValueError, KeyError, TypeError):
        return None


def sender(packet):
    """The identifier the sender of a compact packet names its RPC by."""
    return int.from_bytes(packet[2:6], "big") if len(packet) >= 6 else None


def header_of(packet):
    """The compact header of a packet, its bytes before the payload of a DATA."""
    try:
        decoded = compact.decode(packet)["packet"]
    except compact.DecodeError:
        return packet
    return packet[:len(packet) - len(decoded["payload"]) // 2] if decoded["type"] == "DATA" else packet


def marks(c):
    """Where the stamps of the central stand, the next entry of its reads and of its writes."""
    return len(c.reads), len(c.writes)


def draw_stamps(c, mark, large, quick, **row):
    """One draw with the stamps since `mark` of the packets of its two RPCs, each value read as
    [header, read, hand out] and each write as [header, made, written], the header in hex. The
    client names an RPC by its identifier and the server by the client's with the low bit set."""
    ids = {large.rpc.local, large.rpc.local ^ 1, quick.rpc.local, quick.rpc.local ^ 1}
    reads = [[header_of(v).hex(), r, h] for r, h, v in c.reads[mark[0]:] if sender(v) in ids and h is not None]
    writes = [[header_of(p).hex(), m, w] for m, w, p in c.writes[mark[1]:] if sender(p) in ids and w is not None]
    return dict(row, large=large.rpc.local, small=quick.rpc.local, reads=reads, writes=writes)


def wait_64(c, mark, quick):
    """The wait of section 6.4 in ms, from the write at its event of the one packet of the small
    request to the hand out of the last packet of its answer, from the stamps of the central."""
    written = next(w for m, w, p in c.writes[mark[1]:] if sender(p) == quick.rpc.local and p[0] == compact.DATA)
    handed = max(h for r, h, v in c.reads[mark[0]:] if sender(v) == quick.rpc.local ^ 1 and v[0] == compact.DATA)
    return (handed - written) / 1e6


def plan():
    """The seeded shares of the download of the large answer at which the draws of section 6.4
    send the small request, the same in both columns."""
    rng = random.Random(SPEC["seed"])
    return [rng.uniform(0, 1) for _ in range(SPEC["wait_draws"])]


def phase_graphtp():
    c, t, e, info = central(None)
    OUT["central"] = info
    client = g.Client(e)
    values = []
    for n in range(SPEC["operations"]):
        length = RNG.randint(1, 1024)
        values.append("".join(RNG.choice(KANJI) for _ in range(max(1, length // 3))) if n % 10 == 9
                      else "".join(RNG.choice(ALPHABET) for _ in range(length)))
    ok, flying, pending = 0, [], collections.deque(values)
    deadline = time.monotonic() + 300
    while (pending or flying) and time.monotonic() < deadline:
        while pending and len(flying) < 4:
            v = pending.popleft()
            flying.append((v, client.operation(echo_content(v), repeatable=True)))
        e.poll(0.002)
        left = []
        for v, ex in flying:
            if ex.done:
                ok += v == echoed(client.take(ex))
            else:
                left.append((v, ex))
        flying = left
    OUT["operations"] = {"operations": len(values), "ok": ok}
    small = b'{"query":"{hello}"}'
    big_value = "".join(RNG.choice(ALPHABET) for _ in range(16384))
    counts = t.counts
    client.run(client.operation(small, repeatable=True), timeout=30)
    before = dict(counts)
    s = client.run(client.operation(small, repeatable=True), timeout=30)
    after_small = dict(counts)
    began = time.monotonic()
    b = client.run(client.operation(echo_content(big_value), repeatable=True), timeout=120)
    transfer = time.monotonic() - began
    after_big = dict(counts)
    # The form the record of the wait measured: the small operation sent when
    # the first packet of the large answer is handed out, timed from its
    # making with its uplink.
    record_form, record_draws = [], []
    for _ in range(SPEC["draws"]):
        mark = marks(c)
        large = client.operation(echo_content(big_value), repeatable=True)
        while large.rpc.state == "out":
            e.poll(0.001)
        left = time.monotonic()
        quick = client.operation(small, repeatable=True)
        client.wait([quick], timeout=60, step=0.001)
        record_form.append(1e3 * (time.monotonic() - left))
        client.wait([large], timeout=120, step=0.001)
        record_draws.append(draw_stamps(c, mark, large, quick, wait_ms=record_form[-1]))
    # The calibration draw: the download of the large answer, from the hand
    # out of its first packet to that of its last.
    mark = marks(c)
    large = client.operation(echo_content(big_value), repeatable=True)
    client.wait([large], timeout=120, step=0.001)
    handed = [h for r, h, v in c.reads[mark[0]:] if sender(v) == large.rpc.local ^ 1 and v[0] == compact.DATA]
    span = (max(handed) - min(handed)) / 1e9
    # The form of section 6.4: the small request sent at a seeded share of
    # that download after the hand out of the first packet of the large answer.
    section_64, draws_64 = [], []
    for where in plan():
        mark = marks(c)
        large = client.operation(echo_content(big_value), repeatable=True)
        while large.rpc.state == "out":
            e.poll(0.001)
        at = time.monotonic() + where * span
        while time.monotonic() < at:
            e.poll(min(0.001, max(0.0, at - time.monotonic())))
        quick = client.operation(small, repeatable=True)
        client.wait([quick], timeout=60, step=0.001)
        client.wait([large], timeout=120, step=0.001)
        section_64.append(wait_64(c, mark, quick))
        draws_64.append(draw_stamps(c, mark, large, quick, where=where, wait_ms=section_64[-1]))
    OUT["table_ok"] = s.get("body") == b'{"data":{"hello":"world"}}' and echoed(b) == big_value
    OUT["table"] = {"bytes up, the small operation": after_small["link_bytes_out"] - before["link_bytes_out"],
                    "bytes down, its answer": after_small["link_bytes_in"] - before["link_bytes_in"],
                    "bytes down, 16 KiB": after_big["link_bytes_in"] - after_small["link_bytes_in"],
                    "wait behind 16 KiB, ms, section 6.4": sum(section_64) / len(section_64),
                    "wait behind 16 KiB, ms, the record's form": sum(record_form) / len(record_form),
                    "goodput, bytes a second": len(big_value) / transfer}
    OUT["download_ms"] = 1e3 * span
    OUT["waits"] = {"section 6.4": section_64, "the record's form": record_form}
    OUT["draws"] = {"section 6.4": draws_64, "the record's form": record_draws}
    c.disconnect()


def phase_baseline():
    c = paired()[0]
    base = Baseline(c, max(c.att_mtu() - 3, 20))
    small = b'{"query":"{hello}"}'
    big_value = "".join(RNG.choice(ALPHABET) for _ in range(16384))
    big = echo_content(big_value)
    base.send(small)
    base.receive()
    before = dict(base.counts)
    base.send(small)
    s = base.receive()
    after_small = dict(base.counts)
    began = time.monotonic()
    base.send(big)
    b = base.receive()
    transfer = time.monotonic() - began
    after_big = dict(base.counts)
    # The form the record measured, as the column GraphTP(Homa) times it: the
    # small request sent when the first value of the large answer is handed
    # out, timed from its making with its uplink.
    record_form = []
    for _ in range(SPEC["draws"]):
        base.send(big)
        while not base.inbuf:
            base.poll(0.001)
        left = time.monotonic()
        base.send(small)
        base.receive()
        base.receive()
        record_form.append(1e3 * (time.monotonic() - left))
    # The calibration draw: the download of the large answer, from the hand
    # out of its first value to that of the value that completes it.
    base.send(big)
    while not base.inbuf:
        base.poll(0.001)
    first = base.ends[0][1]
    base.receive()
    span = (base.last_ns - first) / 1e9
    # The form of section 6.4, with the shares of the column GraphTP(Homa):
    # the small request of one packet sent at its share of the download after
    # the hand out of the first value of the large answer.
    section_64 = []
    for where in plan():
        base.send(big)
        while not base.inbuf:
            base.poll(0.001)
        at = time.monotonic() + where * span
        while time.monotonic() < at:
            base.poll(0.001)
        mark = len(c.writes)
        base.send(small)
        base.receive()
        base.receive()
        section_64.append((base.last_ns - c.writes[mark][1]) / 1e6)
    OUT["table_ok"] = s == b'{"data":{"hello":"world"}}' and json.loads(b)["data"]["echo"] == big_value
    OUT["table"] = {"bytes up, the small operation": after_small["link_bytes_out"] - before["link_bytes_out"],
                    "bytes down, its answer": after_small["link_bytes_in"] - before["link_bytes_in"],
                    "bytes down, 16 KiB": after_big["link_bytes_in"] - after_small["link_bytes_in"],
                    "wait behind 16 KiB, ms, section 6.4": sum(section_64) / len(section_64),
                    "wait behind 16 KiB, ms, the record's form": sum(record_form) / len(record_form),
                    "goodput, bytes a second": len(big_value) / transfer}
    OUT["download_ms"] = 1e3 * span
    OUT["waits"] = {"section 6.4": section_64, "the record's form": record_form}
    c.disconnect()


try:
    {"limits": phase_limits, "transfer": phase_transfer, "baseline-transfer": phase_baseline_transfer,
     "graphtp": phase_graphtp, "baseline": phase_baseline}[PHASE]()
except Exception as error:
    OUT["error"] = "%s: %s" % (type(error).__name__, error)
with open("/mnt/xfer/%s.json" % PHASE, "w") as handle:
    json.dump(OUT, handle)
'''

PHASE_SCRIPT = """set -u
cd /mnt/xfer
printf '[General]\\nAlwaysPairable = true\\n[GATT]\\nExchangeMTU = {mtu}\\n' > /etc/bluetooth/main.conf
systemctl restart bluetooth
for i in $(seq 100); do busctl --system list 2>/dev/null | grep -q org.bluez && break; sleep 0.1; done
sleep 1
/mnt/tree/target/debug/aosabi-devtool carrier --requests /mnt/xfer/{phase}-device.ndjson > /mnt/xfer/{phase}-device.out 2> /mnt/xfer/{phase}-device.err &
DEVICE=$!
sleep 1
PYTHONPATH=/mnt/tree/tests timeout {timeout} python3 /mnt/xfer/central.py {phase} '{spec}' > /mnt/xfer/{phase}-central.out 2> /mnt/xfer/{phase}-central.err
echo "central exit $?"
for i in $(seq 150); do kill -0 $DEVICE 2>/dev/null || break; sleep 0.1; done
kill -9 $DEVICE 2>/dev/null
wait
journalctl -u bluetooth --no-pager -n 60 > /mnt/xfer/{phase}-journal.txt 2>&1
dmesg | tail -40 > /mnt/xfer/{phase}-dmesg.txt 2>&1
tail -5 /mnt/xfer/{phase}-central.err /mnt/xfer/{phase}-device.err
"""

UUID_TEXT = {"service": "ab46be00-ba01-4a5f-ba66-9fa7080a2ff4", "psm": "ab46be01-ba01-4a5f-ba66-9fa7080a2ff4",
             "rx": "ab46be02-ba01-4a5f-ba66-9fa7080a2ff4", "tx": "ab46be03-ba01-4a5f-ba66-9fa7080a2ff4"}

# The pace of the data path of the driver, which every request of serve names
# and the driver answers: the interval and the packets an event the central
# paces the link at, and the packets below the engine at most.
PACE = {"interval_ms": interval_ms, "per_event": packets_per_event, "tx_depth": tx_depth}


# The smallest ATT_MTU bluetoothd of guest A takes in `ExchangeMTU`.
SMALLEST_ATT_MTU = 48


def phase(name, mtu, device_request):
    """Run one phase in the guest and return what the central and the driver answered."""
    xfer = GUEST.xfer()
    (xfer / "central.py").write_text(GUEST_PROGRAM, encoding="utf-8")
    (xfer / (name + "-device.ndjson")).write_text(json.dumps(dict(device_request, id=1)) + "\n", encoding="utf-8")
    for leftover in (name + ".json", name + "-device.out"):
        (xfer / leftover).unlink(missing_ok=True)
    spec = json.dumps(dict(UUID_TEXT, seed=seed, operations=operations, transfers=transfers, draws=table_draws,
                           wait_draws=wait_draws, interval_ms=interval_ms, per_event=packets_per_event))
    began = time.monotonic()
    result = GUEST.run(PHASE_SCRIPT.format(phase=name, mtu=mtu, spec=spec, timeout=int(timeout_s * 0.8)),
                       timeout=timeout_s, check=False)
    seconds = time.monotonic() - began
    print("phase %s at ATT_MTU %d: %.1f s, %s" % (name, mtu, seconds, result.out.strip()[-1200:]))
    central = json.loads((xfer / (name + ".json")).read_text(encoding="utf-8")) if (xfer / (name + ".json")).is_file() else {}
    text = (xfer / (name + "-device.out")).read_text(encoding="utf-8") if (xfer / (name + "-device.out")).is_file() else ""
    device = json.loads(text.splitlines()[0]) if text.strip() else {}
    if not central or "error" in central:
        PANICS.append("phase %s: the central answered %s" % (name, central.get("error", "nothing")))
    return central, device, seconds


TRACE_CHECKS = []


def trace_check(name, device):
    """What the trace of the driver kept against what its data path counted: every notification
    `send` handed to the bus, and no more values from RX than the central wrote."""
    trace = device.get("trace") or {}
    packets = trace.get("packets") or []
    gatt = device.get("gatt") or (device.get("counts") or {}).get("gatt") or {}
    sends = sum(1 for _, way, _ in packets if way == "send")
    return {"phase": name, "kept": len(packets), "dropped": trace.get("dropped"), "sends": sends,
            "notifications": gatt.get("notifications"), "recvs": len(packets) - sends, "writes": gatt.get("writes")}
```

### The wait per layer

The trace of the driver and the stamps of the central share the realtime clock of guest A, so every
packet of a draw of GraphTP(Homa) carries the stamps of the layers the record names: a packet of the
server its send in the driver, the read of its `PropertiesChanged` by the central, which is its
arrival in `arrived`, and its hand out, and a write of the central its making, its write at its
event and its recv in the driver. The cell joins the stamps of both ends by the compact header of
each packet, in the order the packets crossed, and counts every packet whose stamps are out of order
across the layers. For each draw it counts the packets of the large answer the driver had handed to
the bus and the central had not yet handed out when the driver sent the small answer, split into
those the central had not yet read, which wait on the bus, in `bluetoothd`, in the kernel and in
`btvirt`, and those in `arrived`. It splits the wait into the hop of the small request from the
central's write to the driver's recv, the server from that recv to the send of the answer, and the
queue below the engine from that send to the hand out, splits that queue once more into the time
before the central reads the answer and the time in `arrived`, and prints the uplink of the request
beside them. It also splits the packets of the server in the draws by whether the driver hands the
bus another packet within 10 ms, since `bluetoothd` holds the last notification of a burst for its
timer `NFY_MULT_TIMEOUT` of 10 ms where the central set the bit of Multiple Handle Value
Notification, and counts the draws whose small answer ends its burst. Of the packets of the large
answer the central had not yet read when the driver sent the small answer, it counts those the
driver sent within 2 ms before it, and it holds the most packets of the large answer below the
engine in a draw of section 6.4 to `tx_depth`, the depth of the pace of the driver.

```python
def head(header):
    """The type, the sender's identifier and the offset of the compact header `header` in hex."""
    raw = bytes.fromhex(header)
    part = raw[0] == compact.DATA and not raw[1] & compact.WHOLE
    return raw[0], int.from_bytes(raw[2:6], "big"), (int.from_bytes(raw[10:14], "big") if part else 0)


def driver_times(device):
    """The times of the trace of the driver by way and compact header, in the order the packets crossed."""
    times = {"send": collections.defaultdict(collections.deque), "recv": collections.defaultdict(collections.deque)}
    for at_ns, way, header in (device.get("trace") or {}).get("packets", []):
        times[way][header].append(int(at_ns))
    return times


# A notification of the data path waits in bluetoothd until the next one or for 10 ms,
# NFY_MULT_TIMEOUT, where the central set the bit of Multiple Handle Value Notification, so the
# last packet of a burst waits for the timer.
NFY_MULT_TIMEOUT_NS = 10_000_000
SENDS = sorted(int(at_ns) for at_ns, way, _ in (GRAPHTP_DEVICE.get("trace") or {}).get("packets", [])
               if way == "send")


def ends_burst(sent):
    """Whether the driver hands the bus no other packet within 10 ms after its send at `sent`."""
    after = bisect.bisect_right(SENDS, sent)
    return after == len(SENDS) or SENDS[after] - sent > NFY_MULT_TIMEOUT_NS


def layered(draw, times):
    """The stamps of one draw joined with those of the driver, and its measures per layer.

    A packet of the server is (type, sender, offset, send, read, hand out) and a write of the
    central (type, sender, offset, made, written, recv), each time in ns of the realtime clock and
    None where the trace of the driver holds no such packet."""
    down = [head(h) + (times["send"][h].popleft() if times["send"][h] else None, read, handed)
            for h, read, handed in draw["reads"]]
    up = [head(h) + (made, written, times["recv"][h].popleft() if times["recv"][h] else None)
          for h, made, written in draw["writes"]]
    out = {"joined": False,
           "unmatched": sum(1 for p in down if p[3] is None) + sum(1 for p in up if p[5] is None),
           "disorder": sum(1 for p in down + up if None not in p[3:] and not p[3] <= p[4] <= p[5])}
    small, large = draw["small"], draw["large"]
    request = [p for p in up if p[0] == compact.DATA and p[1] == small]
    answer = [p for p in down if p[0] == compact.DATA and p[1] == small ^ 1]
    if not request or not answer:
        return out
    _, _, _, made, written, recv = request[0]
    _, _, _, sent, read, handed = max(answer, key=lambda p: p[5])
    if recv is None or sent is None:
        return out
    below = [p for p in down if p[0] == compact.DATA and p[1] == large ^ 1 and p[3] is not None
             and p[3] < sent < p[5]]
    on_bus = [p for p in below if p[4] > sent]
    out.update({"joined": True, "below": len(below), "on the bus": len(on_bus), "in arrived": len(below) - len(on_bus),
                "on the bus sent before": [sent - p[3] for p in on_bus],
                "hop": (recv - written) / 1e6, "server": (sent - recv) / 1e6, "queue": (handed - sent) / 1e6,
                "queue before the read": (read - sent) / 1e6, "queue in arrived": (handed - read) / 1e6,
                "wait": (handed - written) / 1e6, "uplink": (written - made) / 1e6,
                "small ends a burst": ends_burst(sent),
                "bursts": [(ends_burst(p[3]), (p[4] - p[3]) / 1e6) for p in down if p[3] is not None]})
    return out


LAYER_TIMES = driver_times(GRAPHTP_DEVICE)
LAYERS = {form: [layered(draw, LAYER_TIMES) for draw in draws]
          for form, draws in (GRAPHTP.get("draws") or {}).items()}
LAYER_DRAWS = sum(len(rows) for rows in LAYERS.values())
LAYER_JOINED = sum(1 for rows in LAYERS.values() for row in rows if row["joined"])
LAYER_UNMATCHED = sum(row["unmatched"] for rows in LAYERS.values() for row in rows)
LAYER_DISORDER = sum(row["disorder"] for rows in LAYERS.values() for row in rows)
print("draws joined with the trace of the driver: %d of %d, packets of the central the trace does not hold: %d,"
      " packets whose stamps are out of order across the layers: %d, MEASURED" % (
          LAYER_JOINED, LAYER_DRAWS, LAYER_UNMATCHED, LAYER_DISORDER))

MEASURES = [
    ("below", "packets of the large answer below the engine when the driver sent the small answer"),
    ("on the bus", "  not yet read by the central: the bus, bluetoothd, the kernel and btvirt"),
    ("in arrived", "  in arrived"),
    ("hop", "the hop of the small request, the central's write to the driver's recv, ms"),
    ("server", "the server, the driver's recv to the send of the answer, ms"),
    ("queue", "the queue below the engine, the send of the answer to its hand out, ms"),
    ("queue before the read", "  before the central reads it: the bus, bluetoothd, the kernel and btvirt, ms"),
    ("queue in arrived", "  in arrived, ms"),
    ("wait", "the wait, the central's write to the hand out, ms"),
    ("uplink", "the uplink, the making of the small request to its write at its event, ms"),
]
LAYER_SUMMARY = {}
for form, rows in LAYERS.items():
    joined = [row for row in rows if row["joined"]]
    LAYER_SUMMARY[form] = {key: statistics.mean(row[key] for row in joined) for key, _ in MEASURES} if joined else {}
    print()
    print("the wait per layer on GraphTP(Homa), %s, %d draws joined of %d, the mean [least, most], MEASURED"
          % (form, len(joined), len(rows)))
    for key, label in MEASURES:
        values = [row[key] for row in joined]
        print("  %-84s %s" % (label, "%.1f [%.1f, %.1f]" % (statistics.mean(values), min(values), max(values))
                              if values else "none"))

BURSTS = [pair for rows in LAYERS.values() for row in rows if row["joined"] for pair in row["bursts"]]
print()
for ends, label in ((True, "end a burst, no other packet sent within 10 ms"), (False, "are followed within 10 ms")):
    values = [ms for end, ms in BURSTS if end == ends]
    print("packets of the server in the draws that %s: %d, from the send in the driver to the central's read %s,"
          " MEASURED" % (label, len(values), "median %.1f ms [%.1f, %.1f]" % (
              statistics.median(values), min(values), max(values)) if values else "none"))
SMALL_ENDS = {form: sum(1 for row in rows if row["joined"] and row["small ends a burst"])
              for form, rows in LAYERS.items()}
print("the small answer ends its burst in %s, MEASURED" % ", ".join(
    "%d of %d draws of %s" % (SMALL_ENDS[form], len(rows), form) for form, rows in LAYERS.items()))

# The packets of the large answer the central had not yet read when the
# driver sent the small answer, counted by whether the driver sent them within
# 2 ms before it.
WITHIN_NS = 2_000_000
ON_BUS_BEFORE = {form: [ns for row in rows if row["joined"] for ns in row["on the bus sent before"]]
                 for form, rows in LAYERS.items()}
print("packets of the large answer not yet read by the central when the driver sent the small answer, sent within"
      " 2 ms before it: %s, MEASURED" % ", ".join(
          "%d of %d in the draws of %s" % (sum(1 for ns in before if ns <= WITHIN_NS), len(before), form)
          for form, before in ON_BUS_BEFORE.items()))

# The packets of the answer a window of the GATT profile lets below the engine: the central grants
# a window of 8 of its packets beyond the bytes it holds in order, and the packet that starts at
# the edge of a grant goes out whole, so at most the window and one payload less a byte, in
# packets of the answer, the largest packet the data path sends less the 14 bytes of the header of
# a part.
REFERENCE = gatt_reference(185)
PAYLOAD = REFERENCE["sent_packet"] - 14
WINDOW_PACKETS = -(-(REFERENCE["window"] + PAYLOAD - 1) // PAYLOAD)
RECORD = LAYER_SUMMARY.get("the record's form") or {}
FORM_64 = LAYER_SUMMARY.get("section 6.4") or {}
JOINED_64 = [row for row in LAYERS.get("section 6.4", []) if row["joined"]]
MOST_BELOW = max((row["below"] for rows in LAYERS.values() for row in rows if row["joined"]), default=0)
BELOW_64 = sum(row["below"] for row in JOINED_64)
ARRIVED_64 = sum(row["in arrived"] for row in JOINED_64)
MOST_BELOW_64 = max((row["below"] for row in JOINED_64), default=None)
print()
print("packets of the large answer below the engine when the driver sent the small answer, at most %s in a draw"
      " of section 6.4 against the depth of the pace of the driver, %d, and at most %d in any draw, MEASURED"
      % (MOST_BELOW_64, tx_depth, MOST_BELOW))


def held(ok):
    """The state of a cause of the record against its measure."""
    return "held" if ok else "refuted"


CAUSES = []
if RECORD and FORM_64:
    CAUSES = [
        ("packets of the large answer wait below the engine when the small answer is ready, more than the 2 the"
         " design note counts, the record's section The likely cause",
         held(FORM_64["below"] > 2),
         "%.1f packets in the form of section 6.4 and %.1f in the form the record measured, held where the first"
         " exceeds 2" % (FORM_64["below"], RECORD["below"])),
        ("about 10 of them, the lead's reading in issue #1 and the arithmetic of 150 ms as 5 connection events of"
         " 2 packets",
         held(7.5 <= RECORD["below"] <= 12.5),
         "%.1f packets in the form the record measured, of whose %.1f ms from the making of the request %.1f ms"
         " are its uplink, held where the packets lie within a quarter of 10" % (
             RECORD["below"], RECORD["uplink"] + RECORD["wait"], RECORD["uplink"])),
        ("the window of the engine on the GATT profile, 4 unscheduled packets and a window of 8, bounds the"
         " packets below the engine, the record's first cause",
         held(MOST_BELOW <= WINDOW_PACKETS),
         "at most %d packets in any draw, held where no draw holds more than the %d packets of the answer a window"
         " of %d bytes lets through" % (MOST_BELOW, WINDOW_PACKETS, REFERENCE["window"])),
        ("the queue of the emulated link is the central's, arrived, behind whose packets the small answer waits,"
         " the record's second cause",
         held(FORM_64["queue in arrived"] > FORM_64["queue before the read"]),
         "of %.1f ms of queue below the engine in the form of section 6.4 the small answer spends %.1f ms in"
         " arrived and %.1f ms before the central reads it, and %.1f of the %.1f packets ahead of it are in"
         " arrived when the driver sends it, held where arrived holds the larger part of the time" % (
             FORM_64["queue"], FORM_64["queue in arrived"], FORM_64["queue before the read"],
             FORM_64["in arrived"], FORM_64["below"])),
        ("the form the record measured times the wait where the queue below the engine has just filled, the"
         " record's third cause",
         held(RECORD["below"] >= FORM_64["below"]),
         "%.1f packets below the engine in that form against %.1f in the form of section 6.4, beside %.1f ms of"
         " uplink in that form, held where the first is at least the second" % (
             RECORD["below"], FORM_64["below"], RECORD["uplink"])),
    ]
print()
print("the causes the record assumes, each against the measure above, MEASURED")
for cause, state, measure in CAUSES:
    print("  %s: %s, %s" % (state.upper(), cause, measure))
print("  NOT JUDGED HERE: on real controllers the queue moves below user space, the record's fourth cause, which"
      " the bench of DC12 measures")
```
